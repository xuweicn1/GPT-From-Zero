# Flask Demo

下面是一个简单的 Flask 演示应用：

```python
from flask import Flask, jsonify, request

app = Flask(__name__)

@app.route('/')
def index():
    return 'Hello, Flask!'

@app.route('/api/echo', methods=['POST'])
def echo():
    data = request.get_json()
    return jsonify({'you_sent': data})

if __name__ == '__main__':
    app.run(debug=True)
```

说明：
- `/` 路由返回简单文本。
- `/api/echo` 路由接收 JSON 请求并返回相同内容。
