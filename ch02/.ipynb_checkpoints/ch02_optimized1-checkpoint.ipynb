# 第二章 处理文本数据

## 2.2 文本分词

分词工具参考：https://tiktokenizer.vercel.app/?model=gpt2

```python
import os
import urllib.request
import re
import torch
import tiktoken
from torch.utils.data import Dataset, DataLoader

url = 'https://raw.githubusercontent.com/xuweicn1/GPT-From-Zero/refs/heads/ch02/ch02/the-verdict.txt'
file_path = 'the-verdict.txt'

if not os.path.exists(file_path):
    urllib.request.urlretrieve(url, file_path)

with open(file_path, 'r', encoding='utf-8') as f:
    raw_text = f.read()

print(f"文本长度: {len(raw_text)} 字符")
```

### 简单分词示例

```python
text = 'Hello, world. This, is a test.'
result = re.split(r'(\s)', text)
print(result)
```

### 改进分词规则

```python
# 增加标点符号作为分隔符
result = re.split(r'([,.]|\s)', text)
result = [item.strip() for item in result if item.strip()]
print(result)
```

### 更复杂的分词器

```python
text = 'Hello, world. Is This-- is a test?'
result = re.split(r'([,.:;?_!"()\']|--|\s)', text)
result = [item.strip() for item in result if item.strip()]
print(result)
```

### 应用到完整文本

```python
processed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
processed = [item.strip() for item in processed if item.strip()]
print(f"词元数量: {len(processed)}")
print(f"前30个词元:\n{processed[:30]}")
```

## 2.3 将词元转换为词元 ID

```python
all_words = sorted(set(processed))
vocab_size = len(all_words)
print(f"词汇表大小: {vocab_size}")

# 构建词汇表：词元 -> ID
vocab = {token: idx for idx, token in enumerate(all_words)}
```

### 实现简单分词器

```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {v: k for k, v in vocab.items()}

    def encode(self, text):
        processed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        processed = [item.strip() for item in processed if item.strip()]
        return [self.str_to_int[s] for s in processed]

    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
tokenizer = SimpleTokenizerV1(vocab)
text = '"It\'s the last he painted, you know," Mrs. Gisburn said with pardonable pride.'
ids = tokenizer.encode(text)
print(f"词元ID: {ids}")
print(f"解码结果: {tokenizer.decode(ids)}")
```

## 2.4 特殊上下文词元处理

- `<|endoftext|>`：表示文本结束或分隔不同文档
- `<|unk|>`：表示未知词元

```python
all_tokens = sorted(set(processed))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])
vocab = {token: idx for idx, token in enumerate(all_tokens)}
print(f"扩展后词汇表大小: {len(vocab)}")
```

### 改进后的分词器

```python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i: s for s, i in vocab.items()}

    def encode(self, text):
        processed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        processed = [item.strip() for item in processed if item.strip()]
        processed = [item if item in self.str_to_int else "<|unk|>" for item in processed]
        return [self.str_to_int[s] for s in processed]

    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
tokenizer = SimpleTokenizerV2(vocab)
text = "Hello, do you like tea?"
print(f"编码: {tokenizer.encode(text)}")
print(f"解码: {tokenizer.decode(tokenizer.encode(text))}")
```

### 使用分隔符连接文本

```python
text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palace."
text = " <|endoftext|> ".join((text1, text2))
print(f"编码结果: {tokenizer.encode(text)}")
```

## 2.5 BPE 分词器（以 GPT-2 为例）

```python
tokenizer = tiktoken.get_encoding("gpt2")
print(f"'Hello world' 编码: {tokenizer.encode('Hello world')}")
print(f"解码: {tokenizer.decode([15496, 995])}")
```

```python
text = "Hello, do you like tea? <|endoftext|> In the sunlit terraces of someunknownPlace."
integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
print(f"编码: {integers}")
print(f"解码: {tokenizer.decode(integers)}")
```

## 2.6 使用滑动窗口进行数据采样

```python
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()

enc_text = tokenizer.encode(raw_text)
print(f"编码后长度: {len(enc_text)}")
```

### 示例：生成输入-目标对

```python
enc_sample = enc_text[50:]
context_size = 4

x = enc_sample[:context_size]
y = enc_sample[1:context_size+1]

print(f"x: {x}")
print(f"y: {y}")
```

### 构建数据集和数据加载器

```python
class GPTDatasetV1(Dataset):
    def __init__(self, txt, tokenizer, max_length, stride):
        self.input_ids = []
        self.target_ids = []
        token_ids = tokenizer.encode(txt, allowed_special={"<|endoftext|>"})
        for i in range(0, len(token_ids) - max_length, stride):
            self.input_ids.append(torch.tensor(token_ids[i:i+max_length]))
            self.target_ids.append(torch.tensor(token_ids[i+1:i+max_length+1]))

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, idx):
        return self.input_ids[idx], self.target_ids[idx]

def create_dataloader_v1(txt, batch_size=4, max_length=256, stride=128,
                         shuffle=True, drop_last=True, num_workers=0):
    tokenizer = tiktoken.get_encoding("gpt2")
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)
    return DataLoader(dataset, batch_size=batch_size, shuffle=shuffle,
                      drop_last=drop_last, num_workers=num_workers)
```

### 测试数据加载器

```python
dataloader = create_dataloader_v1(raw_text, batch_size=1, max_length=4, stride=1, shuffle=False)
data_iter = iter(dataloader)
first_batch = next(data_iter)
print(f"第一批次: {first_batch}")
```

### 避免数据重叠（设置 stride = max_length）

```python
dataloader = create_dataloader_v1(raw_text, batch_size=8, max_length=4, stride=4, shuffle=False)
inputs, targets = next(iter(dataloader))
print("Inputs:\n", inputs)
print("\nTargets:\n", targets)
```

## 2.7 创建词元嵌入

```python
vocab_size = 6
output_dim = 3

torch.manual_seed(123)
embedding_layer = torch.nn.Embedding(vocab_size, output_dim)
print(f"嵌入权重:\n{embedding_layer.weight}")

input_ids = torch.tensor([2, 3, 5, 1])
print(f"\n输入 ID 的嵌入:\n{embedding_layer(input_ids)}")
```

## 2.8 编码词位置

GPT-2 使用绝对位置嵌入

```python
vocab_size = 50257
output_dim = 256
max_length = 4

token_embedding_layer = torch.nn.Embedding(vocab_size, output_dim)
pos_embedding_layer = torch.nn.Embedding(max_length, output_dim)

dataloader = create_dataloader_v1(raw_text, batch_size=8, max_length=max_length,
                                  stride=max_length, shuffle=False)
inputs, targets = next(iter(dataloader))

token_embeddings = token_embedding_layer(inputs)
pos_embeddings = pos_embedding_layer(torch.arange(max_length))
input_embeddings = token_embeddings + pos_embeddings

print(f"输入嵌入形状: {input_embeddings.shape}")
```

## 小结

本章完成了从原始文本到模型可处理的嵌入向量的完整流程：

1. 文本分词（正则 + BPE）
2. 构建词汇表与分词器
3. 处理未知词元和文本分隔符
4. 滑动窗口采样训练数据
5. 词元嵌入 + 位置嵌入
