# 第二章 处理文本数据

欢迎来到第二章！本章我们将学习如何把原始文本转换成大语言模型能够理解的数字格式。

## 2.2 文本分词

在开始之前，我们先理解什么是'分词'。简单来说，就是把一段文字拆分成一个个小片段（我们称之为'词元'或 token）。

> 在线体验分词：[tiktokenizer 可视化工具](https://tiktokenizer.vercel.app/?model=gpt2)

```python
import os
import urllib.request  

# 下载文本文件（如果本地还没有的话）
url = 'https://raw.githubusercontent.com/xuweicn1/GPT-From-Zero/refs/heads/ch02/ch02/the-verdict.txt'
file_path = 'the-verdict.txt'

if not os.path.exists('the-verdict.txt'):
    urllib.request.urlretrieve(url, file_path)

# 读取文件内容
with open(file_path, 'r') as f:
    raw_text = f.read()

print(f"文本长度: {len(raw_text)} 个字符")
```

```python
# 我们的目标：把这段文本拆分成有意义的片段
import re

# 先用一个简单的例子试试手
text = 'Hello, world. This, is a test.'
result = re.split(r'(\s)', text)
print("按空白字符拆分:", result)
```

```python
# 改进：同时按标点和空白拆分
text = 'Hello, world. This, is a test.'
result = re.split(r'([,.]|\s)', text)
print("按标点和空白拆分:", result)
print("\n问题：出现了很多空字符串")
```

```python
# 加入更多标点符号支持
text = 'Hello, world. Is This-- is a test?'
result = re.split(r'([,.:;?_!"()\']|--|\s)', text)
print("更完善的分词:", result)
```

```python
# 清理：去掉空字符串和两端空白
result = [item.strip() for item in result if item.strip()]
print("清理后:", result)
print("\n✅ 完美！每个词和标点都是独立的元素")
```

```python
# 对整篇小说进行分词
processed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
processed = [item.strip() for item in processed if item.strip()]

print(f"词元总数: {len(processed)}")
print(f"前30个词元: {processed[:30]}")
```

## 2.3 将词元转换为词元 ID

计算机只认识数字，所以我们需要给每个不同的词元分配一个唯一的数字编号（ID）。

```python
# 获取所有不重复的词元并排序
all_words = sorted(set(processed))
vocab_size = len(all_words)
print(f"词汇表大小: {vocab_size}")
```

```python
# 创建词元到ID的映射字典
vocab = {token: integer for integer, token in enumerate(all_words)}

# 查看前5个词元及其ID
for i, (token, idx) in enumerate(vocab.items()):
    if i < 5:
        print(f"{token:15} -> {idx}")
    else:
        break
```

```python
# 实现简单的分词器
class SimpleTokenizerV1:
    def __init__(self, vocab):
        """
        vocab: 字符串到整数的映射字典
        """
        self.str_to_int = vocab  # 文本 -> 数字
        self.int_to_str = {v: k for k, v in vocab.items()}  # 数字 -> 文本

    def encode(self, text):
        """将输入文本转换为整数列表"""
        processed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        processed = [item.strip() for item in processed if item.strip()]
        ids = [self.str_to_int[s] for s in processed]
        return ids

    def decode(self, ids):
        """将整数列表转换回文本"""
        text = " ".join([self.int_to_str[i] for i in ids])
        # 修复标点符号前的空格（英文标点通常紧贴前一个词）
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text

print("✅ 分词器类定义完成")
```

```python
# 测试分词器
tokenizer = SimpleTokenizerV1(vocab)

text = '"It\'s the last he painted, you know," Mrs. Gisburn said with pardonable pride.'
ids = tokenizer.encode(text)
print(f"编码结果: {ids}")
print(f"解码结果: {tokenizer.decode(ids)}")
```

## 2.4 处理特殊词元（未知词和分隔符）

上面的分词器有一个致命问题：遇到词汇表中没有的词会直接报错！

```python
# 尝试编码一个包含新词的句子
text = "Hello, do you like tea"
try:
    print(tokenizer.encode(text))
except KeyError as e:
    print(f"❌ KeyError: 词汇表中没有 '{e}' 这个词！")
```

```python
# 解决方法：添加特殊标记
all_tokens = sorted(list(set(processed)))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])
vocab = {token: integer for integer, token in enumerate(all_tokens)}

print(f"新词汇表大小: {len(vocab)}")
print(f"最后5个词元:")
for item in list(vocab.items())[-5:]:
    print(f"  {item[0]:20} -> {item[1]}")
```

```python
# 改进版分词器 V2
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i: s for s, i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        # 关键：遇到不认识的词，用 <|unk|> 代替
        preprocessed = [
            item if item in self.str_to_int 
            else "<|unk|>" for item in preprocessed
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
        
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text

tokenizer = SimpleTokenizerV2(vocab)
print("✅ V2 分词器创建成功")
```

```python
# 测试 V2 分词器
text = "Hello, do you like tea"
encoded = tokenizer.encode(text)
print(f"编码结果: {encoded}")
print(f"解码结果: {tokenizer.decode(encoded)}")
print("\n✅ 未知词 'Hello' 被成功替换为 <|unk|>")
```

```python
# <|endoftext|> 的用法：拼接两段文本
text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palace."
text = " <|endoftext|> ".join((text1, text2))

print(f"拼接后文本: {text}")
print(f"编码结果: {tokenizer.encode(text)}")
```

## 2.5 BPE 分词器（GPT-2 使用的分词方式）

真正的 GPT 模型使用 BPE（Byte Pair Encoding）算法，能自动发现最优分词方式。我们用 `tiktoken` 库来体验。

```python
import tiktoken

print(f"tiktoken 版本: {tiktoken.__version__}")

# 获取 GPT-2 的预训练分词器
tokenizer = tiktoken.get_encoding("gpt2")

# 测试基本功能
print(f"'Hello world' 编码: {tokenizer.encode('Hello world')}")
print(f"解码回来: {tokenizer.decode([15496, 995])}")
```

```python
# 处理包含特殊标记的文本
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlit terraces"
    "of someunknownPlace."
)

# allowed_special 参数允许特定标记不被拆分
integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})

print(f"编码结果: {integers}")
print(f"解码结果: {tokenizer.decode(integers)}")
```

## 2.6 使用滑动窗口生成训练数据

训练语言模型需要'输入-目标'数据对：给定前面的词，预测下一个词。

```python
# 读取文本并编码
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()

enc_text = tokenizer.encode(raw_text)
print(f"编码后总长度: {len(enc_text)}")
```

```python
# 滑动窗口示例
enc_sample = enc_text[50:]
context_size = 4

x = enc_sample[:context_size]  # 输入：前4个词元
y = enc_sample[1:context_size+1]  # 目标：后4个词元（每个都是前一个的下一个词元）

print(f"输入 (x): {x}")
print(f"目标 (y): {y}")
print("\n每个位置的对应关系:")
for i in range(1, context_size+1):
    context = enc_sample[:i]
    desired = enc_sample[i]
    print(f"  {str(context):30} ----> {desired}")
```

```python
import torch
from torch.utils.data import Dataset, DataLoader

print(f"PyTorch 版本: {torch.__version__}")

class GPTDatasetV1(Dataset):
    def __init__(self, txt, tokenizer, max_length, stride):
        self.input_ids = []
        self.target_ids = []

        # 将整本书编码为数字序列
        token_ids = tokenizer.encode(txt, allowed_special={"<|endoftext|>"})
        assert len(token_ids) > max_length, "文本太短！"

        # 用滑动窗口切分成多个训练样本
        for i in range(0, len(token_ids) - max_length, stride):
            input_chunk = token_ids[i:i + max_length]
            target_chunk = token_ids[i + 1: i + max_length + 1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, idx):
        return self.input_ids[idx], self.target_ids[idx]

print("✅ 数据集类定义完成")
```

```python
def create_dataloader_v1(txt, batch_size=4, max_length=256, 
                         stride=128, shuffle=True, drop_last=True,
                         num_workers=0):
    """
    创建数据加载器
    
    参数:
        txt: 训练文本
        batch_size: 每批样本数
        max_length: 每个样本的词元数
        stride: 滑动窗口步长（步长越小，样本越多但重叠也越多）
        shuffle: 是否打乱数据
        drop_last: 是否丢弃最后不完整的一批
        num_workers: 数据加载进程数
    """
    tokenizer = tiktoken.get_encoding("gpt2")
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)
    
    dataloader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        drop_last=drop_last,
        num_workers=num_workers
    )
    return dataloader

print("✅ 数据加载器函数定义完成")
```

```python
# 测试数据加载器（batch_size=1, max_length=4, stride=1）
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()

dataloader = create_dataloader_v1(
    raw_text, batch_size=1, max_length=4, stride=1, shuffle=False
)

data_iter = iter(dataloader)
first_batch = next(data_iter)
print(f"第一批数据: {first_batch}")
print(f"\n第二批数据: {next(data_iter)}")
print("\n⚠️ 注意：stride=1 时样本高度重叠，容易导致过拟合")
```

```python
# 更好的设置：stride = max_length，样本不重叠
dataloader = create_dataloader_v1(
    raw_text, batch_size=8, max_length=4, stride=4, shuffle=False
)

data_iter = iter(dataloader)
inputs, targets = next(data_iter)
print("输入 (Inputs):")
print(inputs)
print(f"形状: {inputs.shape}")
print("\n目标 (Targets):")
print(targets)
print(f"形状: {targets.shape}")
print("\n✅ 8个样本，每个4个词元，互不重叠")
```

## 2.7 创建词元嵌入

神经网络只能处理数字向量，所以需要把词元 ID 转换成固定长度的向量——这个过程叫'嵌入'。

```python
# 嵌入层的基本使用
input_ids = torch.tensor([2, 3, 5, 1])

vocab_size = 6       # 词汇表有6个词
output_dim = 3       # 每个词用3维向量表示

torch.manual_seed(123)
embedding_layer = torch.nn.Embedding(vocab_size, output_dim)

print("嵌入层的权重矩阵（初始是随机的）:")
print(embedding_layer.weight)
print(f"\n词元 ID=3 对应的向量: {embedding_layer(torch.tensor([3]))}")
```

```python
# 批量查询多个词元ID的向量
embeddings = embedding_layer(input_ids)
print(f"输入词元ID: {input_ids}")
print(f"输出嵌入向量形状: {embeddings.shape}")
print(f"嵌入向量:\n{embeddings}")
print("\n💡 每行对应一个词元ID的向量表示")
```

## 2.8 编码词位置

词的位置信息同样重要！'我打你'和'你打我'词相同但意思不同。GPT 通过添加位置嵌入来解决这个问题。

```python
# 模拟 GPT-2 的嵌入设置（简化版）
vocab_size = 50257
output_dim = 256

# 词元嵌入层
token_embedding_layer = torch.nn.Embedding(vocab_size, output_dim)
print("✅ 词元嵌入层创建完成")
```

```python
# 获取一批训练数据
max_length = 4
dataloader = create_dataloader_v1(
    raw_text, batch_size=8, max_length=max_length,
    stride=max_length, shuffle=False
)
data_iter = iter(dataloader)
inputs, targets = next(data_iter)

print("词元 ID:")
print(inputs)
print(f"形状: {inputs.shape}")
```

```python
# 1. 获取词元嵌入
token_embeddings = token_embedding_layer(inputs)
print(f"词元嵌入形状: {token_embeddings.shape}")
# 含义：8个样本 x 4个词元 x 256维向量
```

```python
# 2. 创建位置嵌入层
context_length = max_length
pos_embedding_layer = torch.nn.Embedding(context_length, output_dim)

# 获取位置嵌入（位置 0,1,2,3 的向量）
pos_embeddings = pos_embedding_layer(torch.arange(max_length))

print(f"位置嵌入形状: {pos_embeddings.shape}")
print("位置嵌入矩阵（4个位置 x 256维）:")
print(pos_embeddings)
```

```python
# 3. 最终输入 = 词元嵌入 + 位置嵌入
# PyTorch 会自动广播 [4, 256] 到 [8, 4, 256]，加到每个样本上
input_embeddings = token_embeddings + pos_embeddings
print(f"最终输入嵌入形状: {input_embeddings.shape}")
print("\n✅ 这就是送入 Transformer 模型的最终输入！")
print("💡 相同位置的词元会加上相同的位置信息，让模型理解词序")
```
