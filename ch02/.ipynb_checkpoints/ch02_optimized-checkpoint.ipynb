# 第二章 处理文本数据

本章将学习如何将原始文本转换为可供 GPT 模型处理的词元（Token）序列。

## 2.1 准备文本数据

首先，我们下载一个示例文本文件（The Verdict），用于后续的分词实验。

```python
import os
import urllib.request
import re

# 下载文本文件（如果不存在）
url = 'https://raw.githubusercontent.com/xuweicn1/GPT-From-Zero/refs/heads/ch02/ch02/the-verdict.txt'
file_path = 'the-verdict.txt'

if not os.path.exists(file_path):
    urllib.request.urlretrieve(url, file_path)

# 读取文件内容
with open(file_path, 'r', encoding='utf-8') as f:
    raw_text = f.read()

print(f"文件共 {len(raw_text)} 个字符")
```

## 2.2 文本分词（Tokenization）

目标：将原始文本分割成单词、标点符号等基本单元（词元）。

```python
# 示例：简单分词
text_sample = 'Hello, world. This, is a test.'
tokens = re.split(r'([,.:;?_!"()\']|--|\s)', text_sample)
tokens = [item.strip() for item in tokens if item.strip()]
print(tokens)
```

```python
# 对整个文本进行分词
preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
preprocessed = [item.strip() for item in preprocessed if item.strip()]

print(f"词元总数: {len(preprocessed)}")
print(f"前30个词元: {preprocessed[:30]}")
```

## 2.3 构建词汇表（Vocabulary）

将唯一的词元映射为整数 ID。

```python
# 获取所有唯一词元并排序
unique_tokens = sorted(set(preprocessed))
vocab_size = len(unique_tokens)

# 创建词元到ID的映射
vocab = {token: idx for idx, token in enumerate(unique_tokens)}

print(f"词汇表大小: {vocab_size}")
```

## 2.4 实现简单的分词器（SimpleTokenizerV1）

该分词器仅支持词汇表中已有的词元。

```python
class SimpleTokenizerV1:
    """简单分词器，不支持未知词元和特殊分隔符"""
    
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {idx: token for token, idx in vocab.items()}
    
    def encode(self, text):
        """将文本编码为词元ID列表"""
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        ids = [self.str_to_int[token] for token in preprocessed]
        return ids
    
    def decode(self, ids):
        """将词元ID列表解码回文本"""
        text = " ".join([self.int_to_str[idx] for idx in ids])
        # 修复标点符号前的空格
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
# 测试 V1 分词器
tokenizer_v1 = SimpleTokenizerV1(vocab)
test_text = '"It\'s the last he painted, you know," Mrs. Gisburn said with pardonable pride.'

ids = tokenizer_v1.encode(test_text)
print("编码结果:", ids)
print("解码结果:", tokenizer_v1.decode(ids))
```

```python
# V1 不支持未知词元，会报错
try:
    print(tokenizer_v1.encode("Hello, do you like tea?"))
except KeyError as e:
    print(f"KeyError: {e} 不在词汇表中")
```

## 2.5 添加特殊词元：`<|unk|>` 和 `<|endoftext|>`

- `<|unk|>`：表示未知词元
- `<|endoftext|>`：表示文本分隔符

```python
# 扩展词汇表
all_tokens = sorted(set(preprocessed))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])
vocab_extended = {token: idx for idx, token in enumerate(all_tokens)}

print(f"扩展后的词汇表大小: {len(vocab_extended)}")
```

## 2.6 支持未知词元的分词器（SimpleTokenizerV2）

```python
class SimpleTokenizerV2:
    """支持未知词元和特殊分隔符的分词器"""
    
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {idx: token for token, idx in vocab.items()}
    
    def encode(self, text):
        """将文本编码为词元ID列表，未知词元替换为<|unk|>"""
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        # 替换未知词元
        preprocessed = [
            item if item in self.str_to_int else "<|unk|>"
            for item in preprocessed
        ]
        ids = [self.str_to_int[token] for token in preprocessed]
        return ids
    
    def decode(self, ids):
        """将词元ID列表解码回文本"""
        text = " ".join([self.int_to_str[idx] for idx in ids])
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
# 测试 V2 分词器
tokenizer_v2 = SimpleTokenizerV2(vocab_extended)
text_unknown = "Hello, do you like tea?"

ids = tokenizer_v2.encode(text_unknown)
print("编码结果:", ids)
print("解码结果:", tokenizer_v2.decode(ids))
```

```python
# 测试文本分隔符
text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palace."
combined_text = f"{text1} <|endoftext|> {text2}"

ids = tokenizer_v2.encode(combined_text)
print("编码结果:", ids)
print("解码结果:", tokenizer_v2.decode(ids))
```

## 2.7 测试与验证

确保编码和解码过程互为逆操作。

```python
def test_tokenizer(tokenizer, test_texts):
    """测试分词器的编码/解码对称性"""
    for i, text in enumerate(test_texts):
        ids = tokenizer.encode(text)
        decoded = tokenizer.decode(ids)
        print(f"测试 {i+1}:")
        print(f"原始: {text}")
        print(f"解码: {decoded}")
        print("-" * 50)

test_cases = [
    "Hello, world!",
    "The price is $100.",
    "This is a test: one, two, three.",
    "<|endoftext|> is a special token."
]

test_tokenizer(tokenizer_v2, test_cases)
```
